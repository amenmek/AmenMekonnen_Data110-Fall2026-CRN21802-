## 1. Introduction
This project analyzes a dataset of 5,000 movies to explore patterns in film runtime, budget , and genre earnings. By examining this we aim to uncover key insights into what leads to sucess in the film industry

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your movie blockbusters dataset
df = pd.read_csv("movie_blockbusters.csv")

# Quick check to make sure it loaded correctly
print(f"Dataset successfully loaded with {len(df)} rows and {len(df.columns)} columns!")
df.head()
```

```python
# Create Histogram for Movie Runtimes
plt.figure(figsize=(9, 5))
plt.hist(df['runtime_min'], bins=25, color='skyblue', edgecolor='black')

plt.title('Distribution of Movie Runtimes', fontsize=14, fontweight='bold')
plt.xlabel('Runtime (Minutes)', fontsize=12)
plt.ylabel('Number of Movies', fontsize=12)
plt.grid(axis='y', alpha=0.3)

plt.show()
```

### Chart 1: Runtime Distribution
- **Key Takeaway:** The majority of movies in this dataset have a runtime between **90 and 130 minutes**, with a strong peak right around 105–110 minutes.
- **Insight:**  studios usually favor a standard 1.5 to 2hour window. Very short movies (< 80 minutes) and very long movies (> 150 minutes) are alot more rarer.

```python
# Create Scatter Plot for Budget vs. Revenue
plt.figure(figsize=(9, 5))
plt.scatter(df['budget_usd'] / 1e6, df['revenue_usd'] / 1e6, alpha=0.5, color='coral', edgecolors='none')

plt.title('Production Budget vs. Worldwide Revenue', fontsize=14, fontweight='bold')
plt.xlabel('Budget (Millions USD)', fontsize=12)
plt.ylabel('Revenue (Millions USD)', fontsize=12)
plt.grid(True, alpha=0.3)

plt.show()
```

### Chart 2 Discussion: Budget vs. Revenue Relationship
- **Key Takeaway:** There is a clear positive relationship between how much a movie costs to make and how much money it brings in at the box office.
- **Insight:** While higher-budget films generally capture higher revenues, spending more money also increases financial risk, as a cluster of high-budget films fail to achieve massive returns.

```python
# Group by genre and calculate average revenue in millions USD
genre_revenue = df.groupby('genre')['revenue_usd'].mean() / 1e6
genre_revenue = genre_revenue.sort_values(ascending=False)

# Create Bar Chart
plt.figure(figsize=(10, 5))
genre_revenue.plot(kind='bar', color='mediumpurple', edgecolor='black')

plt.title('Average Worldwide Revenue by Movie Genre', fontsize=14, fontweight='bold')
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Average Revenue (Millions USD)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.show()
```

### Chart 3 Discussion: Average Revenue by Genre
- **Key Takeaway:** Highstake genres like **Adventure** and **Sci-Fi** average more box office earnings, pulling in way more revenue than genres like Drama or Documentaries.
- **Conclusion:** While lower budget genres can occasionally surprise, studios investing heavily in blockbusters lean into action heavy, globally appealing genres to maximize box office returns.
